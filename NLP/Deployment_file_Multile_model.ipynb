```python
import joblib
```

```python
loaded_model = joblib.load("best_nlp_model.pkl")
```

```python
new_text = ["The government announced a new public transportation project today. Officials said construction will begin next month."]

prediction = loaded_model.predict(new_text)

print(prediction)
```

```python
new_txt = [
    "The national weather department issued a warning for heavy rainfall in several districts and advised residents to follow local safety instructions."
]

prediction_1 = loaded_model.predict(new_txt)

print(prediction_1)
```

```python
new_text = [
    "The government announced a new public transportation project today.",
    "The company reported a 12 percent increase in quarterly revenue.",
    "The weather department issued a warning for heavy rainfall.",
    "The local university announced that its examinations will begin next Monday.",
    "The central bank announced its latest interest rate decision after its policy meeting."
]

predictions = loaded_model.predict(new_text)

for text, prediction in zip(new_text, predictions):
    print("Text:", text)
    print("Prediction:", prediction)
    print("-" * 50)
```
